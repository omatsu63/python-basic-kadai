```python
class Human:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def check_adult(self):

    if self.age >= 20:
      print("大人である")

    else :
      print("大人でない")

human1 = Human("田中", 25)
human2 = Human("松田", 30)
human3 = Human("内藤", 32)
human4 = Human("杉山", 30)
human5 = Human("江藤", 18)

human = [human1, human2, human3, human4, human5]

for i in human:
  i.check_adult()
```
